````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import learning_curve, validation_curve

house = pd.read_csv("Ames_Housing.csv")

features = [
    "OverallQual",
    "GrLivArea",
    "GarageCars",
    "TotalBsmtSF",
    "1stFlrSF",
    "YearBuilt"
]

X = house[features]
y = house["SalePrice"]

print("X 크기:", X.shape)
print("y 크기:", y.shape)

print("\n[결측치]")
display(X.isna().sum().to_frame("missing"))

print("\n[기초 통계량]")
display(X.describe())
````

# 필수 1. 학습곡선으로 과소적합과 과적합 진단

## 배경

학습곡선은 **학습 데이터의 양을 늘려가며 Train score와 Validation score가 어떻게 변하는지 확인하는 그래프**입니다.

````python
import matplotlib.pyplot as plt
import platform

if platform.system() == 'Windows':
    plt.rcParams['font.family'] = 'Malgun Gothic'
elif platform.system() == 'Darwin':  # Mac
    plt.rcParams['font.family'] = 'AppleGothic'

plt.rcParams['axes.unicode_minus'] = False  # 마이너스 기호(-)가 깨지는 것 방지
````

````python
# TODO
# max_depth=2 결정트리의 학습곡선을 그려보세요.
model2 = DecisionTreeRegressor(max_depth=2, random_state=42)

train_sizes, train_scores, valid_scores = learning_curve(model2, X, y, cv = 5, train_sizes=np.linspace(0.1, 1.0, 6))
tr_mean = train_scores.mean(axis=1)
va_mean = valid_scores.mean(axis=1)

result2 = pd.DataFrame({
    'train size': train_sizes,
    'train': tr_mean,
    'valid': va_mean,
    'gap': tr_mean - va_mean
})
display(result2)

plt.figure(figsize=(14,4))
plt.plot(train_sizes, tr_mean, label = 'train')
plt.plot(train_sizes, va_mean, label = 'validation')
plt.title('학습 곡선')
plt.xlabel('train size')
plt.ylabel('score')
plt.legend()
plt.show()
````

# 필수 2. 검증곡선과 train-valid gap으로 최적 복잡도 확인

## 배경

검증곡선은 **모델의 하이퍼파라미터를 바꿔가며 Train score와 Validation score의 변화를 확인하는 그래프**입니다.

이번 문제에서는 결정트리의 `max_depth`를 바꾸어 모델 복잡도를 조절합니다.

## 문제 1. `max_depth` 검증곡선을 만드세요.

### 요구사항

1. 다음 `max_depth` 후보를 사용합니다.

```python
param_range = [1, 2, 3, 4, 5, 8, 12, 20, None]
```

2. `validation_curve()`를 사용합니다.
3. `cv=5`를 사용합니다.
4. 각 `max_depth`의 Train score와 Validation score 평균을 계산합니다.
5. `gap = train_mean - valid_mean`을 계산합니다.
6. `max_depth`, `train`, `validation`, `gap`을 DataFrame으로 만듭니다.
7. Train / Validation score를 하나의 그래프에 표시합니다.

### 확인 포인트

- 복잡도가 증가할수록 Train score가 어떻게 변하는가?
- Validation score는 어느 지점까지 좋아지고 이후 어떻게 변하는가?
- train-valid gap은 깊이가 커질수록 어떻게 변하는가?

````python
# TODO
# validation_curve()를 이용해 max_depth별 Train / Validation score와 gap을 확인하세요.
param_range = [1, 2, 3, 4, 5, 8, 12, 20, None]

train_scores, valid_scores = validation_curve(
    DecisionTreeRegressor(random_state=42),
    X,
    y,
    param_name='max_depth',
    param_range=param_range,
    cv = 5
)

tr_mean = train_scores.mean(axis=1)
va_mean = valid_scores.mean(axis=1)
gap = tr_mean - va_mean

gap_df = pd.DataFrame({
    'max depth': [str(p) for p in param_range],
    'train': tr_mean,
    'validation': va_mean,
    'gap': gap
})
display(gap_df)

plt.figure(figsize=(9, 4))
plt.plot(gap_df['max depth'], gap_df['train'], label='Train')
plt.plot(gap_df['max depth'], gap_df['validation'], label='Validation')
plt.title('validation 곡선')
plt.xlabel('max depth')
plt.ylabel('Score')
plt.legend()
plt.show()

print(f'valid 최고: {gap_df.loc[gap_df['validation'].idxmax(), 'max depth']}')
````

### Q3. Validation score가 가장 높은 `max_depth`는 무엇인가요?

**답변:**  8

### Q4. `max_depth`가 매우 커질수록 train-valid gap이 커지는 이유는 무엇인가요?

**답변:**  깊이가 커지면 트리가 데이터를 계속 분할해 학습 데이터를 거의 외우게 된다. 그래서 트레인 스코어는 거의 1.0까지 올라가지만, 외운 규칙에는 학습 데이터에만 있는 노이즈와 이상치도 들어 있어서 새로운 벨리데이션 데이터에서는 오히려 성능이 떨어진다. 결과적으로 gap이 8에서 0.18, None에서 0.27로 커지고 분산이 커지는 과적합 상태가 된다.
